```python
import rasterio
from rasterio.merge import merge
from rasterio.plot import show
from pathlib import Path

# 1. 타일 폴더 경로 지정
tif_dir = Path(r"D:\LiDAR\FL_Panhandle_B1_2018")

# 2. 모든 .tif 파일 경로 수집
tif_files = list(tif_dir.glob("*.tif"))
print(f"총 {len(tif_files)}개의 타일을 찾았습니다.")

# 3. 각각의 tif 파일 열기
src_files_to_mosaic = [rasterio.open(fp) for fp in tif_files]

# 4. 병합 수행
mosaic, out_transform = merge(src_files_to_mosaic)

# 5. 메타데이터 설정
out_meta = src_files_to_mosaic[0].meta.copy()
out_meta.update({
    "driver": "GTiff",
    "height": mosaic.shape[1],
    "width": mosaic.shape[2],
    "transform": out_transform
})

# 6. 병합된 GeoTIFF 저장
out_fp = tif_dir / "merged_FL_Panhandle_B1_2018.tif"
with rasterio.open(out_fp, "w", **out_meta) as dest:
    dest.write(mosaic)

print(f"병합된 파일이 저장되었습니다: {out_fp}")

# 7. (선택) 시각화
show(mosaic, title="Merged LiDAR Tiles")
```
